```python
rides = ( 
    (101, "Ravi", "Bangalore", 250, 4.5), 
    (102, "Anjali", "Hyderabad", 180, 4.2), 
    (103, "Kiran", "Bangalore", 300, 4.8), 
    (104, "Meena", "Chennai", 220, 4.0), 
    (105, "Arjun", "Bangalore", 150, 3.9) 
) 
print(rides)
 
```

```python
#Extract all rides from Bangalore 
print(rides[0])
print(rides[1])
print(rides[3])
```

```python
# 1. Count total rides
total_rides = len(rides)
print("Total rides:", total_rides)
```

```python
# 2. Extract all rides from Bangalore
bangalore_rides = [ride for ride in rides if ride[2] == "Bangalore"]
print("Bangalore rides:", bangalore_rides)
```

```python
# 3. Total revenue from Bangalore
total_revenue_bangalore = sum(ride[3] for ride in rides if ride[2] == "Bangalore")
print("Total revenue from Bangalore:", total_revenue_bangalore)
```

```python
# 4. Highest rated driver
highest_rated = max(rides, key=lambda x: x[4])
print("Highest rated ride:", highest_rated)
```

```python
# 5. Average rating per city
city_ratings = {}
city_counts = {}

for ride in rides:
    city = ride[2]
    rating = ride[4]
    
    city_ratings[city] = city_ratings.get(city, 0) + rating
    city_counts[city] = city_counts.get(city, 0) + 1
```

```python
# Calculate average
avg_ratings = {}
for city in city_ratings:
    avg_ratings[city] = round(city_ratings[city] / city_counts[city], 1)

print("Average rating per city:", avg_ratings)
```
