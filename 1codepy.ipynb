```python
 #Task 1: Declare the variables and assign the values as above with appropriate Data Type
order_id = "ORD12345"        
customer_name = "Rahul"
product_price = 799.99     
quantity = 3               
discount = 0.10            
is_member = True  
   
```

```python
#Task 2: Identify Data Types
print(type(order_id))
print(type(customer_name))
print(type(product_price))
print(type(quantity))
print(type(discount))
print(type(is_member))     
```

```python
#Task 3: Calculate Total Price
total_price = product_price * quantity
print(total_price)
```

```python
#Task 4: Apply Discount
discount_price = total_price - discount
print(discount_price)
```

```python
 #Add GST (18%)
#Calculate GST on the final price  
#Add it to get the final bill amount
final_price = discount_price + 0.18
print(final_price)
```

```python
#Task 6: Display Output
print(order_id)
print(customer_name)
print(total_price)
print("discount applied:", "yes" if is_member else "no")
print(final_price)


```
