```python
#Data Structures Coding Challenge (Python)
#Problem: Longest Substring Without Repeating Characters
"abcabcbb" 
def longest_substring(s):
    char_set = set()
    left = 0
    max_length = 0

    for right in range(len(s)):
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
        char_set.add(s[right])
        max_length = max(max_length, right - left + 1)

    return max_length

# Input
print(longest_substring("abcabcbb"))  # Output: 3
```

```python
def remove_duplicates(lst):
    seen = set()
    result = []

    for num in lst:
        if num not in seen:
            seen.add(num)
            result.append(num)

    return result

# Input
print(remove_duplicates([1, 2, 2, 3, 4, 3, 5]))
# Output: [1, 2, 3, 4, 5]
```

```python
def find_min_max(tup):
    return min(tup), max(tup)

# Input
print(find_min_max((10, 5, 20, 8)))
# Output: (5, 20)
```

```python
def char_frequency(s):
    freq = {}

    for ch in s:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1

    return freq

# Input
print(char_frequency("programming"))
```

```python
def group_by_length(words):
    result = {}

    for word in words:
        length = len(word)
        if length not in result:
            result[length] = []
        result[length].append(word)

    return result

# Input
print(group_by_length(["cat", "dog", "elephant", "bat", "tiger"]))
```
