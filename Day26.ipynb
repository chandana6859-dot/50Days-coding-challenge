```python
#Find the longest name from tuple
students = ("Anu", "Ravikumar", "Jo") 
print(max(students))
```

```python
#Join Subjects into Single String 
subjects = ("Math", "Science", "English")
result = ", ".join(subjects)

print(result)
```

```python
#Reverse Each Student Name
students = ("Anu", "Ravi")

for name in students:
    print(name[::-1])

```

```python
#Count Vowels in Each Name
students = ("Anu", "Ravi")

for name in students:
    count = 0
    for letter in name.lower():
        if letter in "aeiou":
            count += 1
    print(name, ":", count)
```

```python
#Convert Tuple to Uppercase String 
courses = ("python", "sql", "power bi")

result = ", ".join(course.upper() for course in courses)

print(result)
```
